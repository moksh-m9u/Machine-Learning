```python
import pandas as pd
import seaborn as sns
```

```python
tips= pd.read_csv('tips.csv')
titanic= pd.read_csv('Titanic-Dataset.csv')
iris= pd.read_csv('Iris.csv')
passenger= pd.read_csv('Passenger.csv')
```

```python
passenger[['year', 'month']] = passenger['Month'].str.split('-', expand=True)
passenger['year'] = passenger['year'].astype(int)

passenger['month'] = pd.to_datetime(passenger['Month']).dt.month_name()

passenger = passenger.rename(columns={'#Passengers': 'passengers'})
new_passenger = passenger[['year', 'month', 'passengers']]

new_passenger.head()
```
