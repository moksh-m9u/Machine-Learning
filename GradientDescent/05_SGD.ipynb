```python
class GDRegressor:
    def __init__(self,epochs=100,learning_rate=0.01):
        self.epochs=epochs
        self.learning_rate=learning_rate
        self.coef_= None
        self.intercept_= None

    def fit(self,X_train,y_train):
        # intilize
        self.coef_=np.ones(X_train.shape[1])
        self.intercept_=0
        # update
        for i in range(self.epochs):
            y_pred=np.dot(X_train,self.coef_) + self.intercept_
            der_intercept= -2 * np.mean(y_train-y_pred)
            self.intercept_ = self.intercept_ - (self.learning_rate * der_intercept)


            coef_der= -2 * np.dot(X_train.T,(y_train-y_pred))/X_train.shape[0]
            self.coef_ = self.coef_ - (self.learning_rate * coef_der)
        return self
```

```python
class SGDRegressor:
    def __init__(self,epochs=100,learning_rate=0.01):
        self.epochs=epochs
        self.learning_rate=learning_rate
        self.coef_= None
        self.intercept_= None

    def fit(self,X_train,y_train):
        # intilize
        self.coef_=np.ones(X_train.shape[1])
        self.intercept_=0

        # update
        for i in range(self.epochs):
            for j in range(X_train.shape[0]):
                idx=np.random.randint(0,X_train.shape[0])
                y_hat= np.dot(X_train[idx],self.coef_) + self.intercept_
                intercept_der=-2* (y_train[idx]-y_hat)
                self.intercept_=self.intercept_ - (self.learning_rate * intercept_der)


                coef_der = -2 * X_train[idx] * (y_train[idx] - y_hat)
                self.coef_ = self.coef_ - (self.learning_rate * coef_der)
        return self

    def predict(self,X_test):
        return np.dot(X_test,self.coef_) + self.intercept_
```

```python
sgd=SGDRegressor(1000,0.01)
sgd.fit(X_train,y_train)
mera_y=sgd.predict(X_test)
r2_score(y_test,mera_y)
```
