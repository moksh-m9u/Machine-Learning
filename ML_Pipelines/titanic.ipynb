```python
import pandas as pd
import numpy as np 

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv('Titanic-Dataset.csv')
df.sample(5)
```

```python
df= df.drop(columns=['PassengerId','Name','Ticket','Cabin'])
```

```python
df.head()
```

```python
x_train,x_test, y_train , y_test = train_test_split(df.drop('Survived',axis=1),df['Survived'],test_size=0.3,random_state=42)
```

```python
x_train
```

```python
x_test
```

```python
df.isnull().sum()
```

```python
si_age = SimpleImputer()
si_embarked = SimpleImputer(strategy='most_frequent')
```

```python
x_train_age = si_age.fit_transform(x_train[['Age']])
x_train_embarked_imputed = si_embarked.fit_transform(x_train[['Embarked']])

x_test_age = si_age.transform(x_test[['Age']])
x_test_embarked_imputed = si_embarked.transform(x_test[['Embarked']])
```

```python
x_train_embarked
```

```python
ohe_sex = OneHotEncoder(sparse_output=False,handle_unknown='ignore')
ohe_embarked = OneHotEncoder(sparse_output=False,handle_unknown='ignore')
# One-hot encoding
x_train_sex = ohe_sex.fit_transform(x_train[['Sex']])
x_train_embarked = ohe_embarked.fit_transform(x_train_embarked_imputed)

x_test_sex = ohe_sex.transform(x_test[['Sex']])
x_test_embarked = ohe_embarked.transform(x_test_embarked_imputed)
```

```python
x_train_embarked
```

```python

x_train.head()
```

```python
x_train_rem = x_train.drop(columns=['Sex','Age','Embarked'])
```

```python
x_test_rem = x_test.drop(columns=['Sex','Age','Embarked'])
```

```python
x_train_transformed = np.concatenate((x_train_rem,x_train_age,x_train_sex,x_train_embarked),axis=1)
x_test_transformed = np.concatenate((x_test_rem,x_test_age,x_test_sex,x_test_embarked),axis=1)
```

```python
x_test_transformed.shape
```

```python
clf = DecisionTreeClassifier()
clf.fit(x_train_transformed,y_train)
```

```python
y_pred = clf.predict(x_test_transformed)
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
