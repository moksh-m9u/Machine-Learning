```python
import numpy as np 
import pandas as pd

df = pd.read_csv('customer.csv')
df.sample(5)
```

```python
df = df.iloc[:,2:]
df.sample(4)
```

## Train Test Split

```python
from sklearn.model_selection import train_test_split

# Fixed: Use 'purchased' column as target (y) instead of first two columns
x_train, x_test, y_train, y_test = train_test_split(
    df.drop('purchased', axis=1),  # Features: review, education
    df['purchased'],               # Target: purchased (Yes/No)
    test_size=0.3,
    random_state=42
)

print(f"x_train shape: {x_train.shape}")
print(f"x_test shape: {x_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

# Implementing Ordinal Encoder

We fit our data on x_train and then transform both train and test columns

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']])
oe.fit(x_train)
```

### Transforming

```python
x_train_encoded = oe.transform(x_train)
x_test_encoded = oe.transform(x_test)

print("Original x_train:")
print(x_train.head())
print("\nEncoded x_train:")
print(x_train_encoded[:5])
```

### As we can see labels from 0-2 have been assigned in a hierarchical order

- 0 for poor or school
- 1 for average or ug  
- 2 for good or pg

```python
print("Categories:")
print(oe.categories_)
```

# Implementing Label Encoder

### Label encoding is to be used only for target variable or dependent variable Y and not on independent or input variables 

(as mentioned in their official docs)

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

# Fit and transform the target variable
y_train_encoded = le.fit_transform(y_train)
y_test_encoded = le.transform(y_test)

print("Original y_test:")
print(y_test.head())
print("\nEncoded y_test:")
print(y_test_encoded[:5])
print("\nLabel mapping:")
print(dict(zip(le.classes_, le.transform(le.classes_))))
```
