```python
import numpy as np
import pandas as pd
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.read_csv('covid_toy.csv')
```

```python
df.head()
```

```python
df.sample(9)
```

```python
df["age"].value_counts()
```

```python
def missval(column_name):
    return 100 - df[column_name].value_counts().sum()

for i in range(1, 6):
    col_name = df.columns[i-1]  
    missing = missval(col_name)
    print(f"Column '{col_name}': {missing} missing values")
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(df.drop('has_covid',axis=1), df['has_covid'], test_size=0.3)
print(f"x_train size: {x_train.size}, x_test size: {x_test.size}")
```

```python
# SimpleImputer for fever column
si = SimpleImputer()
x_train_fever = si.fit_transform(x_train[['fever']])
x_test_fever = si.transform(x_test[['fever']])  # Use transform, not fit_transform
print(f"x_train_fever shape: {x_train_fever.shape}")
```

```python
# OrdinalEncoder for cough column
oe = OrdinalEncoder(categories=[['Mild','Strong']])
x_train_cough = oe.fit_transform(x_train[['cough']])
x_test_cough = oe.transform(x_test[['cough']])  # Use transform, not fit_transform
print(f"x_train_cough shape: {x_train_cough.shape}")
```

```python
# OneHotEncoder for gender and city columns - FIX: Add sparse_output=False
ohe = OneHotEncoder(drop='first', sparse_output=False)
x_train_gender_city = ohe.fit_transform(x_train[['gender','city']])
x_test_gender_city = ohe.transform(x_test[['gender','city']])  # Use transform, not fit_transform
print(f"x_train_gender_city shape: {x_train_gender_city.shape}")
```

```python
# Extract age column
x_train_age = x_train.drop(columns=['gender','fever','cough','city']).values
x_test_age = x_test.drop(columns=['gender','fever','cough','city']).values
print(f"x_train_age shape: {x_train_age.shape}")
```

```python
# Concatenate all transformed features
x_train_transformed = np.concatenate((x_train_age, x_train_fever, x_train_gender_city, x_train_cough), axis=1)
x_test_transformed = np.concatenate((x_test_age, x_test_fever, x_test_gender_city, x_test_cough), axis=1)

print(f"x_train_transformed shape: {x_train_transformed.shape}")
print(f"x_test_transformed shape: {x_test_transformed.shape}")
```
