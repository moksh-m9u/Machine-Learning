# Gradient Descent Intuition 
This notebook contains the intution for Gradient Descent and how our model converges to the best fit line even after starting from any random point



we converge to the solution by iterating and changing our coeficients by small value (also called as step size)

```python
import numpy as np
from sklearn.datasets import make_regression
```

using toy dataset
> using just 4 points to understand things better

```python
x,y=make_regression(n_samples=4,n_features=1,n_informative=1,n_targets=1,noise=80,random_state=13)
```

## Keeping slope constant to visualize and understand things better 
Now we will just vary one variable at a time to understand the concept of linear regression in more better way

setting initial random value to b = 100     
   
and we will update b in iterations to converge to our sklearn implementation

```python
#y_pred=((78.35*x)+0).reshape(4)
y_pred=((78.35*x)+100)
```

```python
y_pred.shape
```

### Initial Weights plot when b = 0 

```python
plt.scatter(x,y)
plt.plot(x,lr.predict(x),color='red',label='ols')
plt.plot(x,y_pred,color='green',label='b=0')
plt.legend()
```

```python
m=78.35
b=100
```

### Slope of loss funtion (derivative of loss function )

```python
loss_slope=-2*np.sum(y-m*x.ravel()-b)
```

```python
loss_slope
```

defining the step size

```python
learning_rate=0.1
StepSize=learning_rate*loss_slope
StepSize
```

New intercept

```python
#new intercept
b=b-StepSize
b
```

```python
y_pred1=((78.35*x)+b)
```

```python
plt.scatter(x,y)
plt.plot(x,lr.predict(x),color='red',label='ols')
plt.plot(x,y_pred,color='green',label='b=0')
plt.plot(x,y_pred1,color='pink',label='b=0')
plt.legend()
```

Now we will repeat this process to converge to our solution

```python
loss_slope=-2*np.sum(y-m*x.ravel()-b)
loss_slope
```

```python
StepSize=learning_rate*loss_slope
StepSize
```

```python
b=b-StepSize
b
```

```python
y_pred2=m*x.ravel() + b
```

```python
plt.scatter(x,y)
plt.plot(x,lr.predict(x),color='red',label='ols')
plt.plot(x,y_pred,color='green',label='b=0')
plt.plot(x,y_pred1,color='pink',label='1st iteration')
plt.plot(x,y_pred2,color='yellow',label='2nd iteration')
plt.legend()
```

```python
loss_slope=-2*np.sum(y-m*x.ravel()-b)
loss_slope
```

```python
StepSize=loss_slope*learning_rate
StepSize
```

```python
b=b-StepSize
b
```

```python
y_pred3=m*x.ravel()+b
```

```python
plt.scatter(x,y)
plt.plot(x,lr.predict(x),color='red',label='ols')
plt.plot(x,y_pred,color='green',label='b=0')
plt.plot(x,y_pred1,color='pink',label='1st iteration')
plt.plot(x,y_pred2,color='yellow',label='2nd iteration')
plt.plot(x,y_pred3,color='purple',label='3rd iteration')
plt.legend()
```

Running a loop to converge 

```python
b = -100
m = 78.35
lr = 0.01

epochs = 100

for i in range(epochs):
  loss_slope = -2 * np.sum(y - m*x.ravel() - b)
  b = b - (lr * loss_slope)

  y_pred = m * x + b

  plt.plot(x,y_pred)

plt.scatter(x,y)
b
```
